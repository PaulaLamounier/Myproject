### 1.3 Broadcasting (Regras e Intuição)

O **broadcasting** permite operações entre arrays de shapes diferentes.  
O NumPy “estica” o array menor para que ele tenha o mesmo shape do maior.

**Regras:**
1. Arrays precisam ter o mesmo número de dimensões (o menor recebe dimensões extras de tamanho 1 à esquerda).  
2. Para cada dimensão, os tamanhos devem ser iguais ou um deles deve ser 1.  

**Exemplo:**
- Array x: shape (3, 1)  
- Array y: shape (3,)  

Após broadcasting:  
- `x → (3, 3)`  
- `y → (3, 3)`  

Resultado da operação `x + y`:

### 1.2 Indexação, Fatiamento, Máscaras e Fancy Indexing

O NumPy oferece diferentes formas de acessar e manipular elementos de arrays.  
Essas técnicas são fundamentais para trabalhar com dados de forma eficiente.

---

## 🔹 Indexação Básica

Funciona de forma semelhante às listas do Python.

- `x[0]` → acessa o **primeiro elemento**  
- `x[-1]` → acessa o **último elemento**  
- Em arrays 2D:  
  - `x[1, 2]` → acessa o elemento da **linha 1, coluna 2**  

---

## 🔹 Fatiamento (Slicing)

Permite selecionar subconjuntos do array sem precisar de laços `for`.

- `x[1:4]` → elementos do índice 1 até 3 (exclui o 4)  
- `x[:3]` → do início até o índice 2  
- `x[2:]` → do índice 2 até o final  
- `x[::2]` → elementos com passo 2  

⚡ Importante: fatiamento em NumPy retorna uma **visão (view)** dos dados, não uma cópia.  
Isso significa que alterações no slice refletem no array original.

---

## 🔹 Máscaras Booleanas

Podemos aplicar **condições lógicas** para filtrar elementos.

- `x[x > 5]` → retorna apenas elementos maiores que 5  
- `x[x % 2 == 0]` → retorna apenas os pares  
- Máscaras são arrays booleanos (`True`/`False`) usados para seleção.

✅ Muito útil para filtragem de dados.

---

## 🔹 Fancy Indexing

Permite acessar múltiplos elementos de uma vez usando listas ou arrays de índices.

- `x[[0, 2, 4]]` → retorna os elementos nos índices 0, 2 e 4  
- Em arrays 2D:  
  - `x[[0, 1], [1, 2]]` → seleciona os elementos nas posições (0,1) e (1,2)  

Fancy indexing sempre retorna uma **cópia**, e não uma view.

---

## 📌 Resumo

- **Indexação básica:** acessar elementos individuais  
- **Fatiamento:** extrair subconjuntos contínuos  
- **Máscaras:** seleção condicional baseada em lógica  
- **Fancy indexing:** seleção de múltiplos elementos por listas de índices  

Essas técnicas tornam o NumPy extremamente poderoso para **manipulação de dados vetorizados**.


````python

M = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])

print("Elemento [1, 2]:", M[1, 2])          # 60
print("Duas primeiras linhas:", M[:2, :]) # linhas 0 e 1
print("Duas primeiras colunas:", M[:, :2])

mask = M > 50
print("Elementos > 50 (1D):", M[mask])

rows = [0, 2]; cols = [1, 2]
print("Fancy indexing [0,1] e [2,2]:", M[rows, cols])  # [20, 90]
````

````python
M = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])

print("Elemento [1, 2]:", M[1, 2])      
print("Duas primeiras linhas:", M[:2, :2]) # linhas 0 e 1

    
````

````python
mask = M > 50
print("Elementos > 50 (1D):", M[mask])
````

````python
rows = [0, 2]; cols = [1, 2]
print("Fancy indexing [0,1] e [2,2]:", M[rows, cols])  # [20, 90]

````

### 1.4 Ufuncs, Agregações e Álgebra Linear

---

## 🔹 Funções de Matemática Elementar

Essas funções aplicam operações matemáticas **elemento a elemento** em arrays.  
Elas fazem parte do módulo principal do **NumPy** e são altamente **otimizadas para velocidade**.

- **`np.sqrt(x)`** → Calcula a **raiz quadrada** de cada elemento do array `x`.  
- **`np.log1p(x)`** → Calcula o **logaritmo natural de (1 + x)**.  
  - Mais preciso do que `np.log(1+x)` para valores muito pequenos próximos de zero.  
- **`np.exp(x)`** → Calcula a **exponencial de cada elemento** de `x`, ou seja, \( e^x \).

---

### 📘 Logaritmo (log)

O logaritmo é a **operação inversa da exponenciação**.  
Ele responde à pergunta: *"a que potência devo elevar uma base para obter certo número?"*

**Definição:**  
Se \( b^y = x \), então \( \log_b(x) = y \).

- **Logaritmo Natural (ln):** É o logaritmo com base \( e \) (número de Euler ≈ 2.71828).  
  É o mais comum em **ciências, estatística e engenharia**.

**Exemplo:**  
- \( \log_{10}(100) = 2 \) → pois \( 10^2 = 100 \).  
- Se seu dinheiro **dobra a cada ano**, o logaritmo pode indicar **em quantos anos** você terá uma determinada quantia.

---

### 📘 Exponencial (\( e^x \))

A função exponencial é o **inverso do logaritmo natural**.  
Ela descreve **crescimento ou decaimento contínuo**, proporcional ao valor atual.

**Definição:**  
\( f(x) = e^x \), onde \( e \) ≈ 2.71828.

**O que representa:**  
- Base do **crescimento contínuo**.  
- Exemplo: algo crescendo a uma taxa de 100% por período acelera constantemente → modelado por \( e^x \).

**Aplicações:**  
- Juros compostos contínuos  
- Crescimento populacional  
- Decaimento radioativo  
- Distribuições estatísticas (como a distribuição normal)

---

## 🔹 Funções de Álgebra Linear

O submódulo **`np.linalg`** contém ferramentas essenciais para **álgebra linear**.  
Essas operações são fundamentais em áreas como **machine learning, estatística e física**.

- **`A @ b`** → Produto de matrizes (igual a `np.dot(A, b)`).  
- **`np.linalg.inv(A)`** → Calcula a **matriz inversa** de `A`.  
- **`np.linalg.eig(A)`** → Resolve o problema de **autovalores** e **autovetores**.

---

### 📘 Autovalores e Autovetores

- **Autovalores (eigvals):**  
  Escalares que indicam o "alongamento" ou "encolhimento" de um vetor após transformação pela matriz `A`.

- **Autovetores (eigvecs):**  
  Vetores especiais que **mantêm a mesma direção** após multiplicação pela matriz, mudando apenas de magnitude.  
  Essenciais para entender **transformações lineares** (ex.: PCA em Machine Learning).

---

## 🔹 Funções de Agregação

Essas funções reduzem um array a **um único valor**, resumindo os dados.

- **`x.sum()`** → Soma todos os elementos.  
- **`x.mean()`** → Calcula a média aritmética.  

✅ São fundamentais para **análise estatística** e **resumo de dados**.  
Em vez de usar `loops for` (lentos em Python), o NumPy usa implementações otimizadas em **C**, garantindo **alto desempenho** mesmo em grandes conjuntos de dados.


````python
x = np.array([1, 2, 3, 4], dtype=np.float64)
print("sqrt:", np.sqrt(x))
print("log1p:", np.log1p(x))
print("exp:", np.exp(x))
````

````python
print("sum:", x.sum(), "| mean:", x.mean())
````

````python
A = np.array([[1., 2.],
              [3., 4.]])
b = np.array([1., 0.])

print("A @ b:", A @ b)
print("inv(A):", np.linalg.inv(A))
eigvals, eigvecs = np.linalg.eig(A)
print("autovalores:", eigvals)
print("autovetores:", eigvecs)
````

### 2.2 Seleção, Filtros e SettingWithCopy

### `.loc[]` e `.iloc[]` no Pandas

No Pandas, existem duas formas principais de acessar linhas e colunas em um **DataFrame**:  
- **`.loc[]` → baseado em **rótulos** (nomes de colunas e índices)  
- **`.iloc[]` → baseado em **posições numéricas** (estilo NumPy)

---

## 🔹 `.loc[]` → Seleção por Rótulos

- Usa **nomes de índices e colunas**.  
- Inclui tanto **linhas quanto colunas** especificadas.  

**Exemplos:**  
- `df.loc[2]` → retorna a linha com índice **2**  
- `df.loc[0:3]` → retorna as linhas **0 até 3** (inclusive o 3)  
- `df.loc[:, "coluna1"]` → retorna todos os valores da coluna `"coluna1"`  
- `df.loc[1:4, ["coluna1", "coluna3"]]` → retorna linhas de 1 a 4 das colunas especificadas  

✅ Útil quando você quer trabalhar com **nomes explícitos** do DataFrame.  

---

## 🔹 `.iloc[]` → Seleção por Posições

- Usa **índices numéricos** (como listas/arrays em Python).  
- O intervalo segue a regra do **Python slicing**: o final é **exclusivo**.  

**Exemplos:**  
- `df.iloc[2]` → retorna a terceira linha (posição 2)  
- `df.iloc[0:3]` → retorna as linhas de **0 até 2**  
- `df.iloc[:, 0]` → retorna a **primeira coluna**  
- `df.iloc[1:4, [0, 2]]` → retorna linhas de 1 a 3 das colunas de posição 0 e 2  

✅ Útil quando você não conhece os nomes das colunas, mas sabe a **posição**.  

---

## 📌 Resumo

| Método    | Base de Seleção | Inclusividade | Exemplo |
|-----------|-----------------|---------------|---------|
| `.loc[]`  | Rótulos (nomes) | Intervalo **inclui** o final | `df.loc[0:3, "coluna"]` |
| `.iloc[]` | Posições (índices) | Intervalo **exclui** o final | `df.iloc[0:3, 0]` |

👉 Use `.loc[]` quando quiser trabalhar com **nomes**.  
👉 Use `.iloc[]` quando quiser trabalhar com **posições**.

### Função `pandas.concat`

A função **`pandas.concat`** é usada para **unir DataFrames ou Series** ao longo de um eixo específico.  

👉 Diferente do **`merge`**, o `concat` não procura uma **chave comum** para unir os dados.  
Ele apenas **empilha** ou **cola** os objetos, dependendo do eixo escolhido.

---

## 🔹 Formas de Concatenar

### 1. Empilhar Linhas (`axis=0`)
- **Padrão** no `concat`.  
- Adiciona as linhas de um DataFrame ao final de outro.  
- Útil, por exemplo, para juntar dados de meses diferentes em uma única tabela.  

---

### 2. Adicionar Colunas (`axis=1`)
- Adiciona as colunas de um DataFrame ao lado de outro.  
- O alinhamento é feito pelos **índices**.  
- Útil quando você tem diferentes variáveis para os mesmos registros.  

---

✅ Em resumo, `pandas.concat` é uma ferramenta simples e poderosa para **empilhar ou combinar DataFrames/Series**, sem depender de chaves de junção como no `merge`.


````python
import pandas as pd

df1 = pd.DataFrame({'A': ['A0', 'A1'],
                    'B': ['B0', 'B1']},
                    index=[0, 1])

df2 = pd.DataFrame({'A': ['A2', 'A3'],
                    'B': ['B2', 'B3']},
                    index=[2, 3])


display(df1)
display(df2)    
````

````python
# Concatenação vertical (empilhando linhas)
print("Concatenação Vertical (axis=0):")
concat_vertical = pd.concat([df1, df2], axis=0)
print(concat_vertical)
````

````python
# Concatenação horizontal (adicionando colunas)
df3 = pd.DataFrame({'C': ['C0', 'C1']}, index=[0, 1])
display(df3)

````

````python
print("\nConcatenação Horizontal (axis=1):")
concat_horizontal = pd.concat([df1, df3], axis=1)
print(concat_horizontal)
````

### Diferença entre concat e merge

#### A principal diferença está na lógica da união. O concat é uma união baseada no eixo/índice, enquanto o merge é uma união baseada em valores-chave comuns nas colunas, como em bancos de dados SQL.



### 2.5 Datas, Tempos e Janelas

O Pandas possui funções poderosas para manipulação de **séries temporais**, incluindo criação de intervalos de datas, cálculo de médias móveis e reamostragem de frequências.

---

## 🔹 Funções Principais

### `pd.date_range()`
Gera um intervalo de **datas sequenciais**.

- **start="2024-01-01"** → Data inicial  
- **periods=6** → Número de datas a serem geradas  
- **freq="D"** → Frequência diária  
  - Outras opções incluem:  
    - `"W"` → semanal  
    - `"M"` → mensal  
    - `"H"` → horária  

✅ Muito útil para criar índices de datas em séries temporais.

---

### `.rolling()`
Calcula **médias móveis** (ou outras agregações móveis).  

- **window=3** → Define a “janela” de tempo (3 pontos).  
- A cada passo, a janela se desloca em 1 unidade.  
- Os primeiros pontos retornam **NaN** (não há dados suficientes para completar a janela).  

✅ Usado para suavizar séries temporais e observar tendências.

---

### `.resample()`
Reamostra dados de séries temporais, **alterando a frequência**.

- **"W"** → muda a frequência para **semanal**  
- **`.sum()`** → soma os valores dentro de cada nova janela  

✅ Exemplo: transformar dados diários em totais semanais.  

---

## 📌 Resumo

- `pd.date_range()` → cria intervalos de datas  
- `.rolling(window=n)` → aplica médias/estatísticas móveis  
- `.resample("freq")` → muda a frequência da série temporal e aplica agregações  


````python


rng = pd.date_range("2024-01-01", periods=6, freq="D")
print(rng)
````

````python
s = pd.Series([10, 12, 9, 11, 13, 8], index=rng)
display(s)
````

````python
display(s.rolling(window=4).mean())  # média móvel
````

````python
display(s.resample("W").sum())       # reamostragem semanal
````


## 3. Visualização — Matplotlib (API Orientada a Objetos)

**Containers principais**:

- **Figure**: o *canvas* da figura (arquivo/superfície). Responsável por tamanho, **título global** (`fig.suptitle`), salvar (`fig.savefig`), layout, e por conter um ou mais **Axes**.
- **Axes**: cada **gráfico** dentro da Figure. Chamamos métodos de `ax` para desenhar (`ax.plot`, `ax.scatter`, `ax.hist`...). Eixos X/Y, rótulos, legenda e limites vivem aqui.
- **Artists**: tudo que é desenhado (linhas, textos, marcadores, patches) é um *Artist*.

**Fluxo recomendado (OO)**:
1. `fig, ax = plt.subplots()`
2. Métodos em `ax` para desenhar
3. `ax.set_*` para rótulos/limites; `ax.legend()` quando houver `label`
4. Opcional: `fig.suptitle(...)` e `fig.tight_layout()`

**Stateful (plt.*) vs OO**: o estilo `plt.plot(...)` usa um *current axes*. Para projetos maiores, prefira OO.


### 3.1 Anatomia da API OO

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

display(x)
display(y)  
````

### Quando usar cada gráfico (regra prática objetiva)
- **`plot` (linha)**: séries temporais, funções, tendências contínuas.
- **`scatter` (dispersão)**: relação entre duas variáveis; correlação; outliers.
- **`hist` (histograma)**: distribuição de uma variável; escolha de `bins` importa.
- **`bar` (barras)**: comparar categorias (contagens/estatísticas por grupo).
- **`boxplot`**: comparar distribuições (mediana, quartis, outliers) por grupo.
- **Treemap/árvores**: não há `treeplot` nativo; use `squarify` (treemap), `sklearn.tree.plot_tree` (árvores de decisão) ou `networkx` para grafos/árvores gerais (todos renderizam via Matplotlib).

### Subplots: várias visualizações na mesma Figure
`fig, axes = plt.subplots(nrows, ncols)` cria uma grade. Use `axes[i, j]` (ou `axes.flatten()`).

````python
fig, axes = plt.subplots(2, 2, figsize=(8,6))
axes = axes.flatten()
axes[0].plot(np.linspace(0,1,20), np.linspace(0,1,20))
axes[0].set_title('plot')

axes[1].scatter(np.arange(10), np.arange(10)**0.5)
axes[1].set_title('scatter')

axes[2].hist(rng.normal(0,1,300), bins=15)
axes[2].set_title('hist')

axes[3].boxplot([rng.normal(0,1,200), rng.normal(1,1.5,200)], labels=['G1','G2'])
axes[3].set_title('box')

fig.suptitle('Grid 2x2 com diferentes tipos')
fig.tight_layout()
plt.show()
````

- `ax.grid(True)` adiciona grade.
- `fig.savefig('arquivo.png', dpi=300, bbox_inches='tight')` para exportar.

````python

# Dataset simulado
np.random.seed(42)

dados = pd.DataFrame({
    "preco": np.random.gamma(shape=2.0, scale=80.0, size=300),
    "avaliacao": np.clip(np.random.normal(4.0, 0.5, 300), 1, 5),
    "tipo": np.random.choice(["eletronico", "roupa", "alimento"], size=300, p=[0.4, 0.3, 0.3])
})

''' np.random.gamma()
Esta função é usada para gerar números aleatórios a partir de uma distribuição de probabilidade Gama.

O que faz: Ela simula valores que não podem ser negativos e que tendem a se concentrar em torno de um valor, mas com uma "cauda" longa para valores maiores. Isso é comum em dados como preços, salários ou tempo de espera.

Parâmetros no seu código:

shape=2.0: Determina a forma da distribuição.

scale=80.0: Define a escala dos valores (a média é shape * scale, neste caso, $2.0 * 80.0 = 160.0$).

size=300: O número de valores a serem gerados.

np.clip()
Esta função é usada para limitar (recortar) os valores de um array dentro de um intervalo específico.

O que faz: Se um valor for menor que um limite mínimo, ele se torna o valor mínimo. Se for maior que um limite máximo, ele se torna o valor máximo. Isso é ideal para evitar valores discrepantes ou para garantir que os dados fiquem em um intervalo lógico.

Parâmetros no seu código:

np.random.normal(4.0, 0.5, 300): Primeiro, ele gera 300 números de uma distribuição normal com média 4.0 e desvio padrão 0.5, simulando avaliações.

1: O limite mínimo. Qualquer valor gerado menor que 1 será ajustado para 1.

5: O limite máximo. Qualquer valor gerado maior que 5 será ajustado para 5.

np.random.choice()
Esta função escolhe elementos aleatoriamente de um array, como se você estivesse "sorteando" de uma lista.

O que faz: Ela seleciona amostras de um array unidimensional.

Parâmetros no seu código:

["eletronico", "roupa", "alimento"]: A lista de onde os elementos serão escolhidos.

size=300: O número de escolhas a serem feitas.

p=[0.4, 0.3, 0.3]: Define as probabilidades de cada escolha. Neste caso, eletronico tem 40% de chance de ser escolhido, enquanto roupa e alimento têm 30% cada.  '''
````
